```python
import pandas as pd
import requests, os, folium

# CONFIGURATION
LIVE_RAW_PATH = 'data/raw/master_dataset_real.csv'
VIIRS_URL = "https://firms.modaps.eosdis.nasa.gov/data/active_fire/suomi-npp-viirs-c2/csv/SUOMI_VIIRS_C2_South_Asia_24h.csv"
os.makedirs('data/raw', exist_ok=True)

print("📡 Downloading NASA Live Data...")
try:
    df_live = pd.read_csv(VIIRS_URL)
    # Filter for India Bounding Box
    df_live = df_live[(df_live['latitude'] >= 8) & (df_live['latitude'] <= 37) & 
                      (df_live['longitude'] >= 68) & (df_live['longitude'] <= 97)]
    
    df_live.to_csv(LIVE_RAW_PATH, index=False)
    print(f"✅ Saved {len(df_live)} fires to {LIVE_RAW_PATH}")
except Exception as e:
    print(f"❌ Download failed: {e}")

# Visualization
m = folium.Map(location=[20.59, 78.96], zoom_start=5, tiles='CartoDB dark_matter')
for _, row in df_live.iterrows():
    folium.CircleMarker([row['latitude'], row['longitude']], radius=4, color='red', fill=True).add_to(m)
m.save('data/raw/geospatial_validation_report.html')
print("🗺️ Live Map saved: 'data/raw/geospatial_validation_report.html'")
```

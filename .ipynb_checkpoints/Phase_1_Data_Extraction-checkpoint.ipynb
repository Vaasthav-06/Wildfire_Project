```python
# -----------------------------------------------------------------------------
# STEP 1: INITIALIZE PROJECT ENVIRONMENT
# -----------------------------------------------------------------------------
import os

# Define the local directory structure for data organization
# 'raw' stores original data from APIs; 'processed' will store cleaned data later.
data_directories = ['data/raw', 'data/processed']

for directory in data_directories:
    if not os.path.exists(directory):
        os.makedirs(directory)
        print(f"Directory created: {directory}")
    else:
        print(f"Directory verified: {directory}")

print("Local project environment is ready for data ingestion.")
```

```python
# -----------------------------------------------------------------------------
# STEP 2: DATA ACQUISITION AND WEATHER ENRICHMENT
# -----------------------------------------------------------------------------
import pandas as pd
import numpy as np
import requests
from io import StringIO
from datetime import datetime, timedelta

# Endpoint for NASA VIIRS active fire data (South Asia Regional Feed)
NASA_DATA_SOURCE = "https://firms.modaps.eosdis.nasa.gov/data/active_fire/suomi-npp-viirs-c2/csv/SUOMI_VIIRS_C2_South_Asia_24h.csv"

def fetch_wildfire_data():
    """Retrieves real-time satellite fire data and filters for the India region."""
    try:
        response = requests.get(NASA_DATA_SOURCE, timeout=30)
        response.raise_for_status()
        
        # Load raw CSV data into a dataframe
        raw_df = pd.read_csv(StringIO(response.text))
        
        # Geospatial filter: Geographic bounding box for India
        india_filter = (raw_df['latitude'] >= 8) & (raw_df['latitude'] <= 37) & \
                       (raw_df['longitude'] >= 68) & (raw_df['longitude'] <= 97)
        
        df_localized = raw_df[india_filter].copy()
        
        # Temporal filtering: Isolate data from the last 12 hours
        df_localized['acq_time'] = df_localized['acq_time'].astype(str).str.zfill(4)
        df_localized['timestamp'] = pd.to_datetime(
            df_localized['acq_date'] + ' ' + df_localized['acq_time'], 
            format='%Y-%m-%d %H%M'
        )
        
        time_threshold = datetime.utcnow() - timedelta(hours=12)
        filtered_df = df_localized[df_localized['timestamp'] >= time_threshold].copy()
        
        # Select required features and assign the positive target label (1)
        features = ['latitude', 'longitude', 'acq_date', 'acq_time', 'frp']
        final_fires = filtered_df[features].copy()
        final_fires['fire_detected'] = 1
        
        return final_fires
    except Exception as error:
        print(f"Error during NASA data retrieval: {error}")
        return None

def generate_safe_zones(fire_count):
    """Generates synthetic non-fire coordinates to balance the dataset."""
    sample_size = max(fire_count, 50)
    safe_data = pd.DataFrame({
        'latitude': np.random.uniform(8.0, 37.0, sample_size),
        'longitude': np.random.uniform(68.0, 97.0, sample_size),
        'acq_date': [datetime.utcnow().strftime('%Y-%m-%d')] * sample_size,
        'frp': 0,
        'fire_detected': 0
    })
    return safe_data

# Execute pipeline
print("Accessing NASA thermal anomaly feeds...")
fires_df = fetch_wildfire_data()

if fires_df is not None:
    safe_df = generate_safe_zones(len(fires_df))
    combined_dataset = pd.concat([fires_df, safe_df], ignore_index=True)
    
    # Enrich with weather data (Open-Meteo API)
    print(f"Retrieving meteorological data for {len(combined_dataset)} coordinates...")
    
    weather_url = "https://api.open-meteo.com/v1/forecast"
    weather_list = []
    
    # Processing in batches of 50 to maintain API efficiency
    for i in range(0, len(combined_dataset), 50):
        batch = combined_dataset.iloc[i:i+50]
        params = {
            "latitude": ",".join(batch['latitude'].astype(str)),
            "longitude": ",".join(batch['longitude'].astype(str)),
            "current": "temperature_2m,relative_humidity_2m,wind_speed_10m,soil_moisture_0_to_7cm"
        }
        
        res = requests.get(weather_url, params=params)
        data = res.json()
        
        if isinstance(data, list):
            weather_list.extend([d.get('current', {}) for d in data])
        else:
            weather_list.append(data.get('current', {}))
            
    # Final data consolidation
    weather_df = pd.DataFrame(weather_list)
    final_dataset = pd.concat([combined_dataset, weather_df], axis=1)
    
    # Save the raw master dataset
    final_dataset.to_csv('data/raw/master_dataset_real.csv', index=False)
    print("Master dataset successfully compiled and saved to data/raw/master_dataset_real.csv")
```

```python
# -----------------------------------------------------------------------------
# STEP 3: GEOSPATIAL VALIDATION (LAND MASK FILTERING)
# -----------------------------------------------------------------------------
import folium
from global_land_mask import globe

# Load the master dataset
try:
    df_raw = pd.read_csv('data/raw/master_dataset_real.csv')
    
    # Filter: Ensure coordinates are on landmass to remove ocean artifacts
    df_raw['is_land'] = globe.is_land(df_raw['latitude'], df_raw['longitude'])
    df_final = df_raw[df_raw['is_land']].copy()
    
    print(f"Geospatial Validation: {len(df_final)} points confirmed on land.")
    
    # Initialize interactive map for final verification
    validation_map = folium.Map(location=[20.59, 78.96], zoom_start=5)
    
    for _, row in df_final.iterrows():
        marker_color = 'red' if row['fire_detected'] == 1 else 'green'
        folium.CircleMarker(
            location=[row['latitude'], row['longitude']],
            radius=4,
            color=marker_color,
            fill=True,
            fill_opacity=0.6,
            popup=f"Temp: {row.get('temperature_2m')}C"
        ).add_to(validation_map)
    
    # Save validation report
    validation_map.save('data/raw/geospatial_validation_report.html')
    print("Geospatial validation report generated: data/raw/geospatial_validation_report.html")

except FileNotFoundError:
    print("Error: Master dataset file not found. Ensure Step 2 executed successfully.")
```
